```python
#lower case, distil-roberta distinguishes case
pagevector_lower = [item.lower() for item in pagevector]

#cleantokens

#pagevector_clean = [item.replace('\n',' ') for item in pagevector_clean]
#pagevector_clean = [item.replace('Å‚',' ') for item in pagevector_clean]
#pagevector_clean = [item.replace('Å ','') for item in pagevector_clean]
#pagevector_clean = [item.replace('!','') for item in pagevector_clean]
#pagevector_clean = [item.replace('*','') for item in pagevector_clean]
#pagevector_clean = [item.replace('â„¢','') for item in pagevector_clean]
#pagevector_clean = [item.replace('Å’','') for item in pagevector_clean]
#pagevector_clean = [item.replace('/','') for item in pagevector_clean]
#pagevector_clean = [item.replace('Ëœ','') for item in pagevector_clean]
#pagevector_clean = [item.replace('Ëš','') for item in pagevector_clean]
#pagevector_clean = [item.replace('Å½','') for item in pagevector_clean]
#pagevector_clean = [item.replace('ï¬','') for item in pagevector_clean]
#pagevector_clean = [item.replace('ï¬‚','') for item in pagevector_clean]
#pagevector_clean = [item.replace(';','') for item in pagevector_clean]
#pagevector_clean = [item.replace('Å¡','') for item in pagevector_clean]
#pagevector_clean = [item.replace('incopy notes','') for item in pagevector_clean]
#pagevector_clean = [item.replace('do not edit--changes must be made through file info   correctionkey=','') for item in pagevector_clean]

#clean text

pagevector_clean_no_periods = []
pagevector_clean = []
page = -1
for pages in pagevector_lower:
        page += 1
        #clean_text = clean.clean(pagevector_lower[page], 
        clean_text = clean(pagevector_lower[page], 
        fix_unicode=True, 
        to_ascii=True, 
        lower=True, 
        no_line_breaks=True,
        no_urls=True, 
        no_numbers=False, 
        no_digits=False, 
        no_currency_symbols=True, 
        no_punct=True, 
        replace_with_punct=" ", 
        replace_with_url="<URL>", 
        replace_with_number="<NUMBER>", 
        replace_with_digit="", 
        replace_with_currency_symbol="<CUR>",
        lang='en')
        pagevector_clean_no_periods.append(clean_text)
        
pagevector_clean_no_periods = [item.replace('â„¢','') for item in pagevector_clean_no_periods]       
pagevector_clean_no_periods = [item.replace(',','') for item in pagevector_clean_no_periods]
pagevector_clean_no_periods = [item.replace('-','') for item in pagevector_clean_no_periods]
pagevector_clean_no_periods = [item.replace(';','') for item in pagevector_clean_no_periods]
pagevector_clean_no_periods = [item.replace(':','') for item in pagevector_clean_no_periods]         
        

page = -1        
for pages in pagevector_lower:
        page += 1
        #clean_text = clean.clean(pagevector_lower[page], 
        clean_text = clean(pagevector_lower[page],
        fix_unicode=True, 
        to_ascii=True, 
        lower=True, 
        no_line_breaks=True,
        no_urls=True, 
        no_numbers=False, 
        no_digits=False, 
        no_currency_symbols=True, 
        no_punct=False, 
        replace_with_punct=" ", 
        replace_with_url="<URL>", 
        replace_with_number="<NUMBER>", 
        replace_with_digit="", 
        replace_with_currency_symbol="<CUR>",
        lang='en')
        pagevector_clean.append(clean_text)
        
pagevector_clean = [item.replace('â„¢','') for item in pagevector_clean]       
pagevector_clean = [item.replace(',','') for item in pagevector_clean]
pagevector_clean = [item.replace('-','') for item in pagevector_clean]
pagevector_clean = [item.replace(';','') for item in pagevector_clean]
pagevector_clean = [item.replace(':','') for item in pagevector_clean] 


# In[8]:


pagevector[410]
pagevector_clean[410]
pagevector_clean_no_periods[410]


# In[9]:


#fresh[410]


# In[10]:


def new_week_logic(module_lesson_input):
    week1 = 'lesson ?[1-5]{1}'
    week2 = 'lesson ?[6-9]{1}|lesson 10'
    week3 = 'lesson 1[1-5]{1}'
    intro = 'intro'
    resources = 'resources'
    match_week1 = re.search(week1, module_lesson_input)
    match_week2 = re.search(week2, module_lesson_input)
    match_week3 = re.search(week3, module_lesson_input)
    match_intro = re.search(intro, module_lesson_input)
    match_resources = re.search(resources,module_lesson_input)
    if match_week2 != None: week = "02"
    elif match_week3 != None: week = "03"
    elif match_week1 != None: week = "01"
    elif match_intro != None: week = "intro"
    elif match_resources != None: week = "resources"
    else: week = "NULL"
    return week

module_start_pages = []
page_iterator = 0
text = "module [0-9]+\s*lesson [0-9]+"
resources_text = "resources"

for page in pagevector_clean:
    page_iterator += 1
    if 'text' in page:
            match = re.search(text, page)
            if match != None:
                module_lesson = str(re.findall(text,page,flags = re.I + re.M))
            else: 
                module_lesson = str(module_start_pages[-1][1])
            week = new_week_logic(module_lesson)
            module_week_mapping = [page_iterator, module_lesson,str(week),page]
            module_start_pages.append(module_week_mapping)
    else:  
        if len(module_start_pages) == 0:
              module_lesson = 'Intro'
              week = 'Intro'
              module_start_pages.append([page_iterator,module_lesson,week,page])
        if len(module_start_pages) > 0:
              module_lesson = module_start_pages[-1][1]
              week = module_start_pages[-1][2]
              module_start_pages.append([page_iterator,module_lesson,week,page])
        else:
              module_lesson = 'NULL'
              week = week_logic
              module_start_pages.append([page_iterator,module_lesson,str(week),page])

                
module_start_pages_clean = [item[1].replace(' Å\\n','') for item in module_start_pages]
module_start_pages_clean = [item.replace("'","") for item in module_start_pages_clean]
module_start_pages_clean = [item.replace('[','') for item in module_start_pages_clean]
module_start_pages_clean = [item.replace(']','') for item in module_start_pages_clean]
                
#module_start_pages_clean

from pandas import DataFrame
import numpy as np

module_start_pages_df = DataFrame(module_start_pages, columns=["page","module_lesson","week","text"])
module_start_pages_df['clean_module_lesson'] = module_start_pages_clean
#module_start_pages_df[3] = [item[3].replace('Å','') for item in module_start_pages]

module_start_pages_df["text"].replace('', np.nan, inplace=True)
module_start_pages_df_clean = module_start_pages_df.dropna(subset=["text"])

module_start_pages_df


# In[12]:


module_start_pages_df.to_csv(r'module_start_pages_df.csv')


# In[13]:


### Remove null elements from list of lists


# In[14]:


pagevector_clean_nonulls = [sublist for sublist in pagevector_clean if any(sublist)]
pagevector_clean_no_periods_nonulls = [sublist for sublist in pagevector_clean_no_periods if any(sublist)]

print(len(pagevector_clean_nonulls))


# # Distil-Roberta

# In[15]:


model = SentenceTransformer('paraphrase-distilroberta-base-v1')


# ### Encoding

# In[16]:


page_embeddings = model.encode(pagevector_clean_no_periods_nonulls)
page_embeddings


# In[17]:


for page, embedding in zip(pagevector_clean_no_periods_nonulls, page_embeddings):
    print("Sentence:", page)
    print("Embedding:", embedding)
    print("")


# In[18]:


pagevector_clean_nonulls


# # UMAP

# In[19]:


umap_embeddings = umap.UMAP(n_neighbors=20, 
                            n_components=5, 
                            metric='cosine').fit_transform(page_embeddings)
umap_embeddings
print("min size for lesson =",min(module_start_pages_df_clean.clean_module_lesson.value_counts()))


# # HDBSCAN
# K-means performs best when clusters are:
# 
# - â€œroundâ€ or spherical
# - equally sized
# - equally dense
# - most dense in the center of the sphere
# - not contaminated by noise/outliers

# In[20]:


cluster = hdbscan.HDBSCAN(algorithm='best',min_cluster_size=min(module_start_pages_df_clean.clean_module_lesson.value_counts()),
                          metric='euclidean',                      
                          cluster_selection_method='leaf').fit(umap_embeddings)


# # Results

# In[21]:


# Prepare data
umap_data = umap.UMAP(n_neighbors=module_start_pages_df_clean['clean_module_lesson'].nunique(), n_components=2, min_dist=0.1, metric='cosine').fit_transform(page_embeddings)
#umap_data = umap.UMAP(n_neighbors=4, n_components=2, min_dist=0.1, metric='cosine').fit_transform(page_embeddings)

result = pd.DataFrame(umap_data, columns=['x', 'y'])
result['labels'] = cluster.labels_

# Visualize clusters
fig, ax = plt.subplots(figsize=(20, 10))
outliers = result.loc[result.labels == -1, :]
clustered = result.loc[result.labels != -1, :]
plt.scatter(outliers.x, outliers.y, color='#BDBDBD', s=2)
plt.scatter(clustered.x, clustered.y, c=clustered.labels, s=2, cmap='hsv_r')
plt.colorbar()


# In[22]:


# look at cluster centroid, distance of centroid to module centroid
# module centroid


# In[23]:


fig, ax = plt.subplots(figsize=(20, 10))
cluster.single_linkage_tree_.plot()


#ward linkage may be more granular


# In[24]:


fig, ax = plt.subplots(figsize=(20, 10))
cluster.condensed_tree_.plot(cmap='hsv_r')


# In[25]:


fig, ax = plt.subplots(figsize=(20, 10))
cluster.condensed_tree_.plot(select_clusters=True, cmap='hsv_r')


# In[26]:


result['labels']


# # Validation

# In[27]:


result


# In[28]:


module_start_pages_df_clean['clean_module_lesson']


# In[29]:


#Sentence embedding
#Bert QA


# In[30]:


result


# In[31]:


validate_df = result.join(DataFrame(pagevector_clean_nonulls))
validate_df


# In[32]:


pd.set_option('display.max_colwidth', None)

validate_df[validate_df["labels"]==1]


# In[33]:


validate_df[:100]


# In[34]:


#frequency plot per cluster

#for cluster["labels"] in validate_df:
    # getting the elements frequencies using Counter class
#    elements_count = collections.Counter(cluster["labels"])
    # printing the element and the frequency
#    for key, value in elements_count.items():
#        print(f"{key}: {value}")


# In[ ]:


end = time.time()
print(end - start)
```

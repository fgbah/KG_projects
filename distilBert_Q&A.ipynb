```python
## distilBert Q&A
from transformers import DistilBertTokenizer, DistilBertForQuestionAnswering
import torch

distil_tokenizer = DistilBertTokenizer.from_pretrained('distilbert-base-uncased',return_token_type_ids = True)
distil_model = DistilBertForQuestionAnswering.from_pretrained('distilbert-base-uncased-distilled-squad')
```

```python
from transformers import pipeline 
distil_bert_qa = pipeline('question-answering',model =distil_model,tokenizer=distil_tokenizer)
```

```python
## collect page content and skill description
import pandas as pd
import numpy as np
pal_skills2 = []
pal_content2 = []
page = -1
for text in teachersGuidePal['content_type']:
    page = page+1
    if teachersGuidePal['content_type'][page]== "content":
        pal_content2.append(teachersGuidePal['Skill Description'][page])
    else:
        pal_skills2.append(teachersGuidePal['Skill Description'][page])
```

```python
# concatenate question question words
import pandas as pd
import numpy as np
skills_to_Quest = []
skillnum = -1
concat_1 = "How to "
concat_2 = "?"
for skill in pal_skills2:
    skillnum = skillnum+1
    skills_to_Quest.append(concat_1 + pal_skills2[skillnum] + concat_2)
    
skills_to_Quest  
```

```python
%%time
# checking the q&a output
distil_bert_qa({'context':pal_content2[0],'question':skills_to_Quest[0]})
```

```python
%%time
# main loop to collect scores and answers and convert to pandas DF

import pandas as pd
import numpy as np
page = 0
skillnum = 0
qa_scores = []
qa_answer = [] 
for page in range(0,3287):
    context = pal_content2[page]
    score=[]
    answer=[]
    for skillnum in range(0,20):
        question = skills_to_Quest[skillnum]
        score.append(distil_bert_qa({'context':context,'question':question})['score'])
        answer.append(distil_bert_qa({'context':context,'question':question})['answer'])
        skillnum =skillnum+1
    page =page+1
    qa_scores.append(score)
    qa_answer.append(answer)
        
        
score_df_top_20_distil = pd.DataFrame(qa_scores)
anwer_df_top_20_distil = pd.DataFrame(qa_answer)
```
